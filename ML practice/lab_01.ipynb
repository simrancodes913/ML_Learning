```python
m=list(map(int,input().split(",")))
sum=0
l=len(m)
for i in range(l):
    sum+=i
print("Mean:",sum/l)
```

```python
n=1
for i in m:
    n*=i
print("Geometric mean:",pow(n,1/l))
```

```python
m.sort()
med=m[int(l/2)]
print("median:",med)
```

```python
z=0
for i in m:
   z+=1/i
print("harmonic mean:",l/z) 
```

```python
m=list(map(int,input().split(",")))
ct=dict()
for i in m:
        ct[i]=m.count(i)
print(ct)
print("Mode:",max(ct,key=ct.get))
```

```python
m=list(map(int,input().split(",")))
ct=dict()
for i in m:
        ct[m.count(i)]=i
print(ct)
print("Mode:",ct[max(ct)])
```

```python
sum=0
for i in range(len(m)):
    sum+=m[i]
mean=sum/len(m)
print("Mean:", mean)
for i in range(len(m)):
    var=mean-m[i]/len(m)
print("variance:",var)
print("Standard deviation:",var**0.5)
```

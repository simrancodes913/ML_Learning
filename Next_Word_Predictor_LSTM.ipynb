```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
token=Tokenizer()
token.fit_on_texts([dummy])
```

```python
token.word_index
```

```python
for sentence in dummy.split("\n"):
  print(sentence)
  print(token.texts_to_sequences([sentence]))
  print()
```

```python
"""now make this as supervised learning, take one word as input and res word as output
 let's say someone is writing the world of artificial intelligence..then "the" we can make it as input and for this inout the nextword should be predicted as "world" and then for the input "the world" next word should be predicted as "of" and so on for each sentences """
```

```python
for sentence in dummy.split("\n"):
  tokenized_sentence=token.texts_to_sequences([sentence])[0]
  print(tokenized_sentence)
```

```python
input_sequences=[]
for sentence in dummy.split("\n"):
  tokenized_sentence=token.texts_to_sequences([sentence])[0]
  for i in range(1,len(tokenized_sentence)):
    n_gram=tokenized_sentence[:i+1]
    input_sequences.append(n_gram)
```

```python
input_sequences
```

```python
"""but this has unequal sizes so we will use zero padding...max length of sentence then same number of zeroes pad """
```

```python
maxlen=max(len(x) for x in input_sequences)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
pad=pad_sequences(input_sequences,maxlen=maxlen,padding='pre')
```

```python
pad
```

```python
X=pad[:,:-1]
y=pad[:,-1]
```

```python
X.shape
```

```python
y.shape
```

```python
#so if we are making this as supervised learning then this will be a multiclass-classification problem.
```

```python
## so we will onehot-encode the output column
```

```python
len(token.word_index)
```

```python
from tensorflow.keras.utils import to_categorical
y=to_categorical(y,num_classes=len(token.word_index)+1)##becoz indexing starts with 1 not 0
```

```python
##srchitecture
'''1st layer: EMBEDDING LAYER, 2nd layer: LSTM, 3rd layer: DENSE LAYER'''
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input,Embedding,LSTM,Dense
```

```python
model=Sequential()
model.add(Input(shape=(46,)))
model.add(Embedding(input_dim=177,output_dim=100))
model.add(LSTM(128))
model.add(Dense(177,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.summary()
```

```python
#we add embedding layer converting sparse vectors to numerical vectors.
#in embedding layer each number producces 100 vectors.
#LSTM layer learns sequential steps
#dense & softmax produce possibilities.
```

```python
model.fit(X,y,epochs=100,verbose=1)
```

```python
text="machine"
token_text=token.texts_to_sequences([text])[0]
padding=pad_sequences([token_text],maxlen=46,padding='pre')
```

```python
model.predict(padding).shape
```

```python
import numpy as np
pos=np.argmax(model.predict(padding))
```

```python
for word,index in token.word_index.items():
  if index==pos:
    print(word)
```

```python
text2="what is "
for i in range(6):
  token_text2=token.texts_to_sequences([text2])[0]
  padding2=pad_sequences([token_text2],maxlen=46,padding='pre')
  pos2=np.argmax(model.predict(padding2))
  for word,index in token.word_index.items():
    if index==pos2:
      text2=text2+" "+word
      print(text2)
```

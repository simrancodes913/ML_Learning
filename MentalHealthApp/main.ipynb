### Loading all necessary libraries

```python
import numpy as np
import pandas as pd
```

### Reading the dataset

```python
df=pd.read_csv("survey.csv")
df.head(5)
```

### Removing unnecessary columns

```python
df.drop(['Timestamp', 'state', 'comments'], axis=1,inplace=True)
```

```python
df.head(15)
```

```python
print(type(df))
```

```python
df = df[(df['Age'] >= 18) & (df['Age'] <= 70)]
```

```python
df['Gender'] = df['Gender'].str.lower().str.strip()
```

```python
male_terms = [
    'male','m','man','cis male','cis man','mail','make','mal','male-ish',
    'maile','msle','male (cis)','malr'
]

df['Gender'] = df['Gender'].replace(male_terms, 'male')
```

```python
female_terms = [
    'female','f','woman','female ','cis female','cis-female/femme',
    'femake','femail','female (cis)'
]

df['Gender'] = df['Gender'].replace(female_terms, 'female')
```

```python
df.loc[~df['Gender'].isin(['male','female']), 'Gender'] = 'other'
```

```python
df['Gender'].value_counts()
```

```python
df.describe()
```

### Handling missing values

```python
df.isnull().sum().sum()
```

```python
df.fillna("Unknown",inplace=True)
```

```python
df.isnull().sum().sum()
```

```python
x=df.drop("treatment",axis=1)
y=df["treatment"]
```

```python
for i in x.columns:
    print(i,':','\n',x[i].unique(),'\n')
```

```python
cat_cols=x.select_dtypes(include=["object"]).columns
num_cols=x.select_dtypes(exclude=["object"]).columns
```

### Encoding

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
columns=ColumnTransformer(transformers=[("cat",OneHotEncoder(handle_unknown="ignore"),cat_cols)],remainder="passthrough")
```

```python
x.head(10)
```

### Train-Test-Split the data

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
print(x_train)
```

```python
print(x_test)
```

```python
print(y_train)
```

```python
print(y_test)
```

PIPELINE

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

pipeline = Pipeline([
    ("preprocessor", columns),
    ("model", RandomForestClassifier(
        n_estimators=200,
        max_depth=10,
        min_samples_leaf=5,
        min_samples_split=10,
        random_state=42
    ))
])
```

### Model Training

```python
pipeline.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
train_pred = pipeline.predict(x_train)
test_pred = pipeline.predict(x_test)

print("Train Accuracy:",
      accuracy_score(y_train, train_pred))

print("Test Accuracy:",accuracy_score(y_test, test_pred))
```

```python
y_test.value_counts()
```

```python
print(confusion_matrix(y_test, test_pred))
```

```python
print("Accuracy:", accuracy_score(y_test, test_pred))
```

```python
print(classification_report(y_test, test_pred))
```

```python
print(pipeline.score(x_train, y_train))
print(pipeline.score(x_test, y_test))
```

```python
import joblib
joblib.dump(pipeline,"model.pkl")
```
